```python
bugs_left = 50
print ("you have 50 bug to fix today")


while bugs_left > 0 :
      answer = input ("did you fix any bug today?(y/n)").lower()
      if answer == "y":
         bugs_left = bugs_left - 1
         print (f"great you still have {bugs_left}")
         print ("keep going you have a long day")
      if answer == "n":
         print ("whatare you waiting for")

if bugs_left == 0:
   print ("You don't just solve problems, you dominate them. - Harvey Specter")
```
